### Handling Duplicate Values

First, I will remove the 168 duplicate rows identified earlier to ensure that each observation in the dataset is unique. This is important for accurate analysis and to prevent skewed results, especially when performing aggregations or statistical calculations.

````python
# Remove duplicate rows
merged_df.drop_duplicates(inplace=True)
print(f"Number of duplicate values after removal: {merged_df.duplicated().sum()}")
````

### Handling Missing Values

Next, I will address the missing values in the dataset. Based on the `info()` and `isnull().sum()` outputs, several columns have null entries. I will use appropriate strategies for different data types:

*   **Numerical Columns:** For columns like `seasons`, `imdb_score`, `imdb_votes`, `tmdb_popularity`, and `tmdb_score`, I'll impute missing values with the median. The median is robust to outliers and represents a typical value in the distribution.
*   **Categorical/Object Columns:** For columns such as `description`, `age_certification`, `imdb_id`, `name`, `character`, and `role`, I'll fill missing values with a placeholder string like 'Unknown' or 'Missing'. This allows these rows to be retained for analysis without introducing NaN values into categorical operations.

````python
# Impute missing numerical values with the median
for col in ['seasons', 'imdb_score', 'imdb_votes', 'tmdb_popularity', 'tmdb_score', 'person_id']:
    if col in merged_df.columns:
        merged_df[col].fillna(merged_df[col].median(), inplace=True)

# Impute missing categorical values with 'Unknown' or 'Missing'
merged_df['description'].fillna('No Description', inplace=True)
merged_df['age_certification'].fillna('Unknown', inplace=True)
merged_df['imdb_id'].fillna('Missing', inplace=True)
merged_df['name'].fillna('Unknown', inplace=True)
merged_df['character'].fillna('Unknown', inplace=True)
merged_df['role'].fillna('Unknown', inplace=True)

# Verify that missing values have been handled
print("Missing values after handling:")
print(merged_df.isnull().sum())
````

### What all manipulations have you done and insights you found?

### Data Manipulations Performed:

1.  **Duplicate Removal:** I identified and removed 168 duplicate rows from the `merged_df` DataFrame. This ensures that each record is unique and prevents biases in subsequent analyses.
2.  **Missing Value Imputation for Numerical Columns:** For numerical columns such as `seasons`, `imdb_score`, `imdb_votes`, `tmdb_popularity`, `tmdb_score`, and `person_id`, missing values were imputed using the **median** of their respective columns. The median was chosen as it is less sensitive to outliers compared to the mean, providing a more robust central tendency for imputation.
3.  **Missing Value Imputation for Categorical/Object Columns:** For categorical and object type columns like `description`, `age_certification`, `imdb_id`, `name`, `character`, and `role`, missing values were filled with descriptive placeholder strings ('No Description', 'Unknown', or 'Missing'). This approach retains the rows for analysis while explicitly marking the absence of original data, rather than dropping potentially valuable rows.

### Insights Found:

*   The dataset initially contained **168 duplicate entries**, which have now been successfully removed, leaving a clean dataset for analysis.
*   A significant number of missing values, particularly in `seasons` (~116k), `age_certification` (~68k), `tmdb_score` (~11k), `imdb_id` (~5k), `imdb_score` (~6k), `imdb_votes` (~6k), and `character` (~17k), have been addressed. The dataset is now **completely free of missing values** across all columns.

These manipulations have made the `merged_df` dataset analysis-ready, allowing for more accurate and reliable data visualization and statistical modeling without the complications of missing or redundant data.
