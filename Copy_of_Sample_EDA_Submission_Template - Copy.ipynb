#### Chart - 2

````python
# Calculate number of unique productions per person
productions_per_person = df.groupby('person_id')['id'].nunique().reset_index()
productions_per_person.rename(columns={'id': 'num_unique_productions'}, inplace=True)

# Calculate total number of roles/credits per person
roles_per_person = df.groupby('person_id')['character'].count().reset_index()
roles_per_person.rename(columns={'character': 'total_roles'}, inplace=True)

# Merge the two dataframes to get a single view per person
person_data = pd.merge(productions_per_person, roles_per_person, on='person_id')

plt.figure(figsize=(12, 7))
sns.scatterplot(data=person_data, x='num_unique_productions', y='total_roles', alpha=0.6)
plt.title('Number of Unique Productions vs. Total Roles per Person')
plt.xlabel('Number of Unique Productions')
plt.ylabel('Total Number of Roles/Credits')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?
